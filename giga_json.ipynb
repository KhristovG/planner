````python
def get_embeddings():
    model_name = "sentence-transformers/paraphrase-multilingual-mpnet-base-v2"
    model_kwargs = {'device': 'cpu'}
    encode_kwargs = {'normalize_embeddings': False}
    embedding = HuggingFaceEmbeddings(
        model_name=model_name,
        model_kwargs=model_kwargs,
        encode_kwargs=encode_kwargs
        )
    return embedding

def prep_bd():
    file_path = '/Users/grigoriy/Desktop/work'
    vector_store_loaded = FAISS.load_local(folder_path=file_path, embeddings=get_embeddings(), allow_dangerous_deserialization= True)
    loaded_retriever = vector_store_loaded.as_retriever(search_kwargs={'k': 3}, search_type="mmr")
    return loaded_retriever
````

````python
# Нужно будет придумать логику бинда функций для разных задач, мы же не будем биндить сразу все доступные функции

# Set up the tool
from langchain_core.tools import tool
from langgraph.prebuilt import ToolNode
from langchain_community.tools.tavily_search import TavilySearchResults

@tool
def get_user_cards():
    """Обзор списка доступных пользователю карт. Получение списка карт пользователя"""
    cards_name = ['МИР***1234', 'VISA***2345','MASTERCARD***3456']
    # Логика для получения списка карт пользователя через API или другой механизм.
    return f'Список доступных карт: {cards_name}'

@tool
def block_card(card_name: str)-> str:
    """Block card"""
    # Логика для блокировки карты через API или другой механизм
    return f"Карта {card_name} успешно заблокирована."

@tool
def reissue_card(card_name: str, new_card_type: str = "Momentum")-> str:
    """Перевыпустить карту. Можно выбрать карту 'Momentum' или карту постоянного пользования."""
    # Логика перевыпуска карты
    if new_card_type.lower() == "momentum" or new_card_type.lower() == "моментум":
        return "Вы выбрали карту типа Momentum."
    elif new_card_type.lower() == "постоянного пользования":
        return "Вы быбрали карту постоянного пользования."
    else:
        return "Выберите один из доступных типов карты: 'Momentum' или 'постоянного пользования'."


@tool
def deliver_card(delivery_method: str = "Самокат")-> str:
    """Организовать доставку карты. Можно выбрать способ доставки: 'Самокат' или самовывоз."""
    if delivery_method.lower() == "самокат":
        return "Карта будет доставлена с помощью службы доставки 'Самокат'."
    elif delivery_method.lower() == "самовывоз":
        return "Вы можете забрать карту в пункте самовывоза."
    else:
        return "Выберите один из доступных методов доставки: 'Самокат' или 'Самовывоз'."

@tool
def factual_questions(user_question: str)-> str:
    """Используй search_tool когда нужно ответить на фактологический вопрос."""
    search_tool = TavilySearchResults(
                    max_results=5,
                    # include_answer=True,
                    # include_raw_content=True,
                    # include_images=False,
                    # search_depth="advanced",
                    # include_domains = []
                    # exclude_domains = []
                )
    ans = search_tool.invoke(user_question)
    return ans

@tool
def sberbank_question(bank_question: str)-> str:
    """Ответ на вопросы банковской тематики и касательно продуктов, услуг Сбербанка/Сбера."""

    pf = pd.read_csv('FAQ.csv')

    class StateRAG(TypedDict):
        question: str
        context: List[Document]
        answer: str

    def retrieve_question(state: StateRAG):
        retrieved_docs = prep_bd().invoke(state["question"])
        return {"context": retrieved_docs}
    
    def map_questions_and_answer(state: StateRAG):
        docs_content = "\n\n".join(doc.page_content[8:] for doc in state["context"])
        print(f" Ретривнутые вопросы: \n{docs_content}")
        question = question_choose_agent.invoke({"context": docs_content, "question": state["question"]})
        print(f" Ответ модельки по выбору самого релевантного вопроса: \n{question}")
        # print(question)
        if question.content.lower() not in ['no',"'no'"]:
            # q = pf.loc[pf['Вопрос'] == question.content, 'Ответ'].values[0]
            try:
                # Пытаемся найти ответ с исходным вопросом
                answer = pf.loc[pf['Вопрос'] == question.content, 'Ответ'].values[0]
            except IndexError:
                try:
                    # Если не найдено, пробуем добавить знак вопроса
                    question_with_question_mark = question.content.rstrip('?') + '?'
                    answer = pf.loc[pf['Вопрос'] == question_with_question_mark, 'Ответ'].values[0]
                except IndexError:
                    # Если и это не сработало, пробуем убрать знак вопроса
                    question_without_question_mark = question.content.rstrip('?')
                    answer = pf.loc[pf['Вопрос'] == question_without_question_mark, 'Ответ'].values[0]
            # print(q)
            # answer = ans.invoke({"question": state["question"], "context": q})
            return {'answer':answer}
        return {"answer": 'у меня нет ответа на ваш вопрос'}

    graph_builder = StateGraph(StateRAG).add_sequence([retrieve_question, map_questions_and_answer])
    graph_builder.add_edge(START, "retrieve_question")
    graph = graph_builder.compile()
    return f'{graph.invoke({"question": bank_question})['answer']}'

tools = [get_user_cards, block_card, reissue_card, deliver_card, factual_questions, sberbank_question]#card_balance, transefer_money, 
tool_node = ToolNode(tools)
memory = MemorySaver()
````

````python
import inspect
import json

def get_tool_parameters(tool_name):
    """Извлекает параметры инструмента и возвращает их в виде JSON"""
    
    # Получаем объект функции по имени
    tool = globals().get(tool_name)
    
    if tool is None or not callable(tool):
        raise ValueError(f"Функция с именем '{tool_name}' не найдена.")
    signature = inspect.signature(tool.func)
    parameters = {}
    
    for name, param in signature.parameters.items():
        parameters[name] = {
            "type": param.annotation.__name__ if param.annotation != inspect.Parameter.empty else "Any",
            "required": param.default == inspect.Parameter.empty
        }
    
    return {
        "function_name": tool.name,
        "description": tool.description,
        "parameters": parameters
    }

# def get_tools_parameters(tools):
#     """Возвращает параметры всех инструментов в JSON"""
#     return [get_tool_parameters(tool) for tool in tools]

# # Получение JSON с параметрами
# tools_parameters = get_tools_parameters(tools)
# # Установите ensure_ascii=False для корректного отображения кириллицы
# print(json.dumps(tools_parameters, indent=2, ensure_ascii=False))
````

````python
# Help functions for graph
def add_generated_plan(plan_text, start_mes, completed_items=None, plan_status="start"): # возможно эта функция не нужна, потом через стейт add
    global CACHE_DF

    new_id = CACHE_DF['plan_id'].max() + 1 if not CACHE_DF.empty else 1

    new_row = {
        'plan_id': new_id,
        'plan_text': plan_text,
        'plan_status': plan_status,
        'completed_items': completed_items,
        'start_mes':start_mes,
        'summary':summary.invoke({plan_text}).content
    }

    CACHE_DF = pd.concat([CACHE_DF, pd.DataFrame([new_row])], ignore_index=True)

def display_dataframe_pretty(df, title="Проверяем датафрейм"):
    print("\n" + "-" * 40 + f" {title} " + "-" * 40)
    table = tabulate(df, headers="keys", tablefmt="grid", showindex=False)
    print(table)
    print("-" * (82 + len(title)) + "\n")

def display_result_with_border(result, title="Результат проверки"):
    print("\n" + "-" * 40 + f" {title} " + "-" * 40)
    print(f"{result}")
    print("-" * (80 + len(title)) + "\n")

def extract_relevant_messages(messages):
    cleaned_messages = []
    for message in messages:
        if message.type in ["human", "ai",'tool','tools']:
            # Убираем все после 'SYS:' в контенте AIMessage
            content = message.content#.split("\n\nSYS:")[0]
            cleaned_messages.append(f"{message.type}: {content}")
    return cleaned_messages

# Функция для обновления статуса через агента 
# заменено на механику
# def update_plan_status_with_agent2(df, messages):
#     # Формируем контекст для промпта
#     all_plans = {f"plan_id:{row['plan_id']}, plan_text: {row['plan_text']}" for _, row in df.iterrows()}
#     # all_plans = df[['plan_id','plan_text']].to_string(index=False, header=True)
#     message_context = extract_relevant_messages(messages=messages)

#     # Вызываем агента
#     result = completion_status_agent.invoke(
#         {'all_plans': all_plans,
#          'messages': message_context}
#     )

#     # Извлекаем JSON-строку, удаляя обрамляющие символы

#     print(f"СТАТУС {result.content}")
    
#     content = result.content.strip('```json').strip().strip('```')

#     # Обновляем датафрейм на основе выполненных пунктов
#     plan_updates = eval(content)  # Преобразуем JSON-строку в словарь

#     for id_plan, data in plan_updates.items():
#         plan_id = int(id_plan)  # Преобразуем строковый ключ в число
#         if data.get('plan_status') == 'done':
#             # Если статус done, обновляем план
#             df.loc[df['plan_id'] == plan_id, 'plan_status'] = 'done'
#         if 'completed_items' in data:
#             # Если есть выполненные пункты, обновляем соответствующую колонку
#             df.loc[df['plan_id'] == plan_id, 'completed_items'] = ', '.join(data['completed_items'])

#     return df

#status without agent

def update_plan_status(df, cur_plan_id):
    """
    Обновляет статусы в датафрейме:
    - Все статусы, кроме "done", меняются на "pause".
    - Для строки с plan_id == cur_plan_id устанавливается статус "run".
    - Если количество выполненных пунктов совпадает с количеством пунктов плана,
      статус плана меняется на "done".
    
    :param df: DataFrame с колонками "plan_id", "plan_text", "plan_status", "completed_items".
    :param cur_plan_id: ID плана, который должен быть переведен в статус "run".
    :return: Обновленный DataFrame.
    """
    
    # Преобразуем plan_text в список пунктов плана
    df['plan_items'] = df['plan_text'].apply(lambda x: [item.strip() for item in x.split('\n')])
    
    # Сравниваем количество пунктов плана и выполненных пунктов
    df['is_done'] = df.apply(lambda row: len(row['plan_items']) == len(row['completed_items']), axis=1)
    
    # Устанавливаем статус "done" для планов, где все пункты выполнены
    df.loc[df['is_done'], 'plan_status'] = 'done'
    
    # Обновляем все статусы, кроме "done", на "pause"
    df.loc[df['plan_status'] != 'done', 'plan_status'] = 'pause'
    
    # Устанавливаем статус "run" для строки с текущим планом
    df.loc[df['plan_id'] == cur_plan_id, 'plan_status'] = 'run'
    
    # Удаляем временные колонки
    df.drop(columns=['plan_items', 'is_done'], inplace=True)
    
    return df

def get_curr_plan_id(df, plan):
    # Фильтруем датафрейм по совпадению plan_text с переданным текстом
    matching_plans = df[df['plan_text'] == plan]
    
    # Если совпадений нет, вернуть None
    if matching_plans.empty:
        return None

    # Возврат идентификатора первого совпавшего плана
    return matching_plans['plan_id'].iloc[0]

# Для аутпута тестов
def _print_event(event: dict, _printed: set, max_length=1500):
    current_state = event.get("dialog_state")
    if current_state:
        print("Currently in: ", current_state[-1])
    message = event.get("messages")
    if message:
        if isinstance(message, list):
            message = message[-1]
        if message.id not in _printed:
            msg_repr = message.pretty_repr(html=True)
            if len(msg_repr) > max_length:
                msg_repr = msg_repr[:max_length] + " ... (truncated)"
            print(msg_repr)
            _printed.add(message.id)


def get_next_unfinished_item(plan, df: pd.DataFrame) -> Tuple[Optional[str], Optional[List[str]], Optional[str]]:
    """
    Находит следующий невыполненный пункт плана.
    Возвращает кортеж (plan_id, план, пункт)
    """
    # print('get_next_unfinished_item start')
    matching_plans = df[df['plan_text'] == plan]
    # display_dataframe_pretty(matching_plans)
    for _, row in matching_plans.iterrows():
        plan_id = row['plan_id']
        plan = row['plan_text']
        completed_items = row.get('completed_items', [])
        if completed_items is None:
            completed_items = []
        # Ищем первый невыполненный пункт
        plan_items = [item.strip() for item in plan.split('\n')]
        for item in plan_items:
            # Извлекаем номер пункта (например, "1)" из "1) Описание")
            item_number = item.split(')')[0] + ')'
            # print(f'item_number {item_number}')
            # print(f'completed_items {completed_items}')
            if item_number not in completed_items:
                return plan_id, item_number, item
    
    return None, None, None

def mark_item_completed(updated_df: pd.DataFrame, plan_id: str, item_number: str) -> pd.DataFrame:
    """
    Отмечает пункт как выполненный
    Возвращает обновленный DataFrame
    """
    # Создаем копию DataFrame для безопасной работы
    # updated_df = df.copy()
    
    # Находим строку с соответствующим plan_id
    row_idx = updated_df[updated_df['plan_id'] == plan_id].index[0]
    
    # Обновляем список выполненных пунктов
    completed_items = updated_df.at[row_idx, 'completed_items']
    if completed_items is None:
        completed_items = []
    completed_items.append(item_number)
    updated_df.at[row_idx, 'completed_items'] = completed_items
    
    return updated_df

def extract_function_name(text):
    # Регулярное выражение для поиска названия функции
    pattern = r'(\*\*|`)(\w+)(\*\*|`)'
    
    # Поиск совпадений
    match = re.search(pattern, text)
    
    # Если совпадение найдено, возвращаем имя функции
    if match:
        return match.group(2)  # Захватываем только имя функции
    return None

def execute_item(item, user_mess) -> str:
    """
    Выполняет пункт плана с помощью executor
    """
    # plan_id, item_number, item = get_next_unfinished_item(plan, df)

    # тут регулярку чтобы достать название функции и если нет функции вернуть "зови гигу"
    func = extract_function_name(item)
    # print(func)
    # print(type(func))
    print("===============================МОК Executor=================================")
    if func is not None:
        global func_param
        func_param = get_tool_parameters(func)
        func_param["user_message"] = user_mess
        print(f"Бу, не бойся это я МОКЭКЗЕКУТОРА юзай функцию:\n{json.dumps(func_param, indent=4, ensure_ascii=False)}")
        
        return f"Бу, не бойся это я МОКЭКЗЕКУТОРА юзай функцию:\n{json.dumps(func_param, indent=4, ensure_ascii=False)}"
        # return print(f'зови {func} с параметрами {func_param}')
    giga_json_mock = {
        "function_name": "giga_mock_call",
        "description": "Поиск через Гигу",
        "parameters": {
            "user_question": {
                "type": "str",
                "required": True
            }
        }
    }
    giga_json_mock["user_message"] = user_mess
    print(f"Зови гигу МОК ВЫЗОВА ГИГИ:\n{giga_json_mock}")
    return f"Зови гигу МОК ВЫЗОВА ГИГИ:\n{giga_json_mock}"

# def process_next_item(plan, df: pd.DataFrame, llm: ChatOpenAI) -> Tuple[pd.DataFrame, bool]:
#     """
#     Обрабатывает следующий невыполненный пункт плана
#     Возвращает кортеж (обновленный DataFrame, флаг выполнения)
#     """
#     plan_id, plan, item = get_next_unfinished_item(plan, df)
    
#     if item is None:
#         print("Все пункты планов выполнены")
#         return df, False
    
#     # Извлекаем номер пункта
#     item_number = item.split(')')[0] + ')'
    
#     # Выполняем пункт
#     result = execute_item(llm, item)
    
#     # Отмечаем пункт как выполненный
#     updated_df = mark_item_completed(df, plan_id, item_number)

#     return updated_df, True

import random

def get_status():
    # Генерируем случайное число от 0 до 1
    probability = random.random()
    
    # Если число меньше или равно 0.9, возвращаем "status code 200"
    if probability <= 0.9:
        return "status code 200"
    # Иначе возвращаем "перепланирование\ошибка на строне"
    else:
        return "перепланирование\ошибка на строне"
````
